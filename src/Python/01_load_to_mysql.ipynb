```python
import pandas as pd
from sqlalchemy import create_engine

DB_USER = "root"
DB_PASSWORD = "sjbarad777"
DB_HOST = "localhost"
DB_PORT = 3306
DB_NAME = "paynova_fraud_analytics"

engine = create_engine(
    f"mysql+pymysql://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_NAME}"
)
```

```python
from sqlalchemy import text

with engine.connect() as connection:
    result = connection.execute(
        text("SELECT DATABASE();")
    )

    print("Connected database:", result.scalar())
```

```python
file_path = "E:\my new for the me last for the change the my life for the life only goal\project\Digital Payments Fraud Intelligence\Database\clean\clean.csv"

df_clean = pd.read_csv(
    file_path,
    parse_dates=["timestamp"]
)

print(df_clean.shape)
```

```python
df_clean.to_sql(
    name="raw_upi_transactions",
    con=engine,
    if_exists="replace",
    index=False,
    chunksize=5000
)
```

```python
query = """
SELECT COUNT(*) AS row_count
FROM raw_upi_transactions;
"""

result = pd.read_sql(query, engine)

print(result)
```

```python
query = """
SELECT
    COUNT(*) AS total_rows,
    COUNT(DISTINCT transaction_id) AS unique_transaction_ids
FROM raw_upi_transactions;
"""

pd.read_sql(query, engine)
```
